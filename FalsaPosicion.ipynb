```python
import math


def falsa_posicion(f, a, b, tol=1e-6, max_iter=100):
  """Calcula una aproximacion a la raiz de la funcion f(x) en el intervalo [a,

  b] utilizando el metodo numerico de la Falsa Posicion (Regula Falsi).

  Parametros:
  -----------
  f : function
      Funcion continua a la cual se le busca la raiz f(x) = 0.
  a : float
      Limite inferior del intervalo inicial.
  b : float
      Limite superior del intervalo inicial.
  tol : float, opcional
      Tolerancia del error para el criterio de parada (por defecto es 1e-6).
  max_iter : int, opcional
      Numero maximo de iteraciones permitidas (por defecto es 100).

  Retorna:
  --------
  p : float
      Aproximacion calculada a la raiz de la funcion.
  historial : list
      Lista de tuplas con los datos de cada iteracion:
      (iteracion, a, b, p, f(p), error).
  """
  # Evaluacion de los limites iniciales
  fa = f(a)
  fb = f(b)

  # Verificacion del Teorema de Bolzano (cambio de signo)
  if fa * fb >= 0:
    raise ValueError(
        "El intervalo [a, b] no cumple con la condicion de cambio de signo: "
        "f(a) y f(b) deben tener signos opuestos."
    )

  p_anterior = a
  historial = []

  # Encabezado de la tabla de iteraciones
  print(
      f"{'Iter':<6} | {'a':<12} | {'b':<12} | {'p (Raiz)':<12} | {'f(p)':<12}"
      f" | {'Error':<12}"
  )
  print("-" * 78)

  for i in range(1, max_iter + 1):
    # Formula de la secante para calcular la interseccion p
    p = b - (fb * (b - a)) / (fb - fa)
    fp = f(p)

    # Calculo del error relativo aproximado
    error = abs(p - p_anterior) if i > 1 else abs(b - a)

    # Almacenar datos de la iteracion actual
    historial.append((i, a, b, p, fp, error))

    # Mostrar progreso en consola
    print(
        f"{i:<6d} | {a:<12.8f} | {b:<12.8f} | {p:<12.8f} | {fp:<12.4e} |"
        f" {error:<12.4e}"
    )

    # Criterios de convergencia
    if fp == 0 or error < tol:
      print("-" * 78)
      print(f"Convergencia alcanzada en la iteracion {i}.")
      return p, historial

    # Actualizacion del intervalo conservando el cambio de signo (Biseccion)
    if fa * fp < 0:
      b = p
      fb = fp
    else:
      a = p
      fa = fp

    p_anterior = p

  print("-" * 78)
  print("Advertencia: Se alcanzo el numero maximo de iteraciones.")
  return p, historial


# ==========================================
# Ejemplo de ejecucion
# ==========================================
if __name__ == "__main__":

  # Definicion de la funcion de prueba f(x) = x^3 - 2x - 5
  def funcion_ejemplo(x):
    return x**3 - 2 * x - 5

  # Parametros iniciales
  limite_inferior = 2.0
  limite_superior = 3.0
  tolerancia_deseada = 1e-5
  maximo_iteraciones = 50

  print("Metodo de la Falsa Posicion")
  print(
      f"Funcion: f(x) = x^3 - 2x - 5 en el intervalo [{limite_inferior},"
      f" {limite_superior}]"
  )
  print()

  try:
    raiz, _ = falsa_posicion(
        f=funcion_ejemplo,
        a=limite_inferior,
        b=limite_superior,
        tol=tolerancia_deseada,
        max_iter=maximo_iteraciones,
    )

    print(f"\nResultado final:")
    print(f"Aproximacion a la raiz: x = {raiz:.8f}")
    print(f"Valor de la funcion en la raiz: f(x) = {funcion_ejemplo(raiz):.4e}")

  except ValueError as e:
    print(f"Error de entrada: {e}")
```
